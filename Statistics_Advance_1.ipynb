1. **What is a random variable in probability theory?**
    ## Random Variable

    A **random variable** is a variable that assigns a numerical value to each possible outcome of a random experiment.

    ### Example

    Suppose we toss a coin twice.

    Possible outcomes are:

    `HH, HT, TH, TT`

    Let `X` represent the number of heads.

    Then:

    - `HH → X = 2`
    - `HT → X = 1`
    - `TH → X = 1`
    - `TT → X = 0`

    Here, `X` is a **random variable** because its value depends on the outcome of the random experiment.

    ### Types of Random Variables

    1. **Discrete Random Variable:** Takes countable values.
      - Example: Number of heads when tossing a coin.

    2. **Continuous Random Variable:** Can take any value within a range.
      - Example: Height, weight, or temperature.

2. **What are the types of random variables?**
    ## Types of Random Variables

    There are two main types of random variables:

    ### 1. Discrete Random Variable

    A **discrete random variable** takes a countable number of distinct values.

    **Examples:**
    - Number of heads when tossing a coin.
    - Number of students in a class.
    - Number of defective products.

    Example: If X represents the number of heads when tossing a coin twice:

    `X = 0, 1, 2`

    ### 2. Continuous Random Variable

    A **continuous random variable** can take any value within a given range.

    **Examples:**
    - Height of a person.
    - Weight of a person.
    - Temperature.
    - Time taken to complete a task.

    For example, a person's height could be:

    `170.1 cm, 170.15 cm, 170.157 cm, ...`

    ### Difference

    | Discrete | Continuous |
    |---|---|
    | Countable values | Any value within a range |
    | Usually obtained by counting | Usually obtained by measuring |
    | Example: Number of students | Example: Height of students |

3. **Explain the difference between discrete and continuous distributions.**

    A probability distribution describes how probabilities are assigned to the possible values of a random variable.

    | Discrete Distribution | Continuous Distribution |
    |---|---|
    | Used for discrete random variables | Used for continuous random variables |
    | Has countable, separate values | Can take any value within a range |
    | Probability is assigned to individual values | Probability is assigned to intervals |
    | Uses a **Probability Mass Function (PMF)** | Uses a **Probability Density Function (PDF)** |
    | Example: Number of heads in coin tosses | Example: Height, weight, temperature |

    ### Example of Discrete Distribution

    If X is the number of heads when tossing a coin twice:

    `X = 0, 1, 2`

    ### Example of Continuous Distribution

    If X represents the height of a person, it can take values such as:

    `170.1 cm, 170.15 cm, 170.157 cm, ...`

    There are infinitely many possible values within a range.

4. **What is a binomial distribution, and how is it used in probability?**
    ## Binomial Distribution

    A **Binomial Distribution** is a probability distribution that describes the number of successes in a fixed number of independent trials, where each trial has only two possible outcomes: **success or failure**.

    ### Conditions

    A binomial distribution is used when:

    - There is a fixed number of trials (`n`).
    - Each trial has two possible outcomes.
    - The trials are independent.
    - The probability of success (`p`) remains constant.

    ### Formula

    `P(X = k) = C(n, k) × p^k × (1-p)^(n-k)`

    Where:

    - `n` = Number of trials
    - `k` = Number of successes
    - `p` = Probability of success
    - `1-p` = Probability of failure

    ### Example

    Suppose a coin is tossed **5 times** and we want to find the probability of getting exactly **3 heads**.

    Here:

    - `n = 5`
    - `k = 3`
    - `p = 0.5`

    So, the binomial distribution can be used to calculate the probability of getting exactly 3 heads.

    ### Uses in Probability

    Binomial distribution is commonly used to calculate the probability of a specific number of successes.

    **Examples:**
    - Number of heads in coin tosses.
    - Number of defective products in a batch.
    - Number of successful sales calls.
    - Number of customers who make a purchase.


5. **What is the standard normal distribution, and why is it important?**
    ## Standard Normal Distribution

    The **Standard Normal Distribution** is a normal probability distribution with:

    - **Mean (μ) = 0**
    - **Standard deviation (σ) = 1**

    It has a **bell-shaped and symmetric curve**, with most values concentrated around 0.

    ### Z-Score

    A standard normal distribution uses a **z-score** to represent how many standard deviations a value is away from the mean.

    `z = (x - μ) / σ`

    For example, if a dataset has a mean of 50 and a standard deviation of 10, a value of 70 has:

    `z = (70 - 50) / 10 = 2`

    This means 70 is **2 standard deviations above the mean**.

    ### Why is it Important?

    - Helps compare values from different datasets.
    - Makes normally distributed data easier to analyze.
    - Allows us to calculate probabilities using z-tables or statistical software.
    - Used in hypothesis testing and confidence intervals.
    - Helps identify unusually high or low observations.

6. **What is the Central Limit Theorem (CLT), and why is it critical in statistics?**
    ## Central Limit Theorem (CLT)

    The **Central Limit Theorem (CLT)** states that when we take sufficiently large random samples from a population, the distribution of the **sample means** tends to become approximately normal, regardless of the original population's distribution.

    ### Key Points

    - The samples should be randomly selected and independent.
    - As the sample size increases, the distribution of sample means becomes more approximately normal.
    - The mean of the sampling distribution is approximately equal to the population mean.
    - The standard deviation of the sampling distribution is called the **standard error**.

    ### Example

    Suppose the heights of people in a population do not follow a perfectly normal distribution.

    If we repeatedly take random samples of 30 people and calculate the mean height of each sample, the distribution of these sample means will tend to look approximately like a normal distribution.

    ### Why is CLT Important?

    - Makes statistical analysis possible even when the population distribution is unknown or not normal.
    - Forms the foundation for **hypothesis testing**.
    - Helps construct **confidence intervals**.
    - Allows us to make conclusions about a population using sample data.
    - Widely used in surveys, experiments, business analytics, and scientific research.

7. **What is the significance of confidence intervals in statistical analysis?**
    ## Confidence Intervals

    A **Confidence Interval (CI)** is a range of values used to estimate an unknown population parameter, such as the population mean, from sample data.

    For example, a 95% confidence interval of `(70, 80)` means that the method used to calculate the interval would capture the true population parameter in about 95% of repeated samples.

    ### Significance of Confidence Intervals

    - Shows the **uncertainty** associated with a sample estimate.
    - Provides a range instead of relying on a single value.
    - Helps determine how **precise** an estimate is.
    - A **narrower interval** generally indicates greater precision.
    - A **wider interval** indicates more uncertainty.
    - Helps in statistical decision-making and research.
    - Commonly used in hypothesis testing and scientific studies.

    ### Example

    Suppose a survey estimates that the average customer satisfaction score is 7.5, with a 95% confidence interval of:

    `7.2 to 7.8`

    This indicates that the population mean is estimated to lie within this range using the chosen confidence-interval method.

8. **What is the concept of expected value in a probability distribution?**
    ## Expected Value in a Probability Distribution

    The **Expected Value** is the long-run average value of a random variable when a random experiment is repeated many times.

    It is calculated by multiplying each possible outcome by its probability and adding all the results.

    ### Formula

    `E(X) = Σ [x × P(X = x)]`

    Where:

    - `x` = Possible outcome
    - `P(X = x)` = Probability of that outcome

    ### Example

    Suppose a game has the following outcomes:

    | Outcome | Probability |
    |---|---|
    | ₹10 | 0.5 |
    | ₹20 | 0.3 |
    | ₹50 | 0.2 |

    Expected value:

    `E(X) = (10 × 0.5) + (20 × 0.3) + (50 × 0.2)`

    `E(X) = 5 + 6 + 10 = ₹21`

    So, the expected value is **₹21**.

    ### Importance

    - Helps find the average outcome over many repetitions.
    - Useful for decision-making under uncertainty.
    - Used in finance, insurance, business, and risk analysis.
    - Helps compare different probability-based outcomes.

9. **Write a Python program to generate 1000 random numbers from a normal
distribution with mean = 50 and standard deviation = 5. Compute its mean and standard deviation using NumPy, and draw a histogram to visualize the distribution.**



```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 random numbers from a normal distribution
data = np.random.normal(loc=50, scale=5, size=1000)

# Compute mean and standard deviation
mean = np.mean(data)
std_dev = np.std(data)

print("Mean:", mean)
print("Standard Deviation:", std_dev)

# Draw histogram
plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, edgecolor="black")
plt.title("Normal Distribution (Mean = 50, SD = 5)")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

10. ** You are working as a data analyst for a retail company. The company has
collected daily sales data for 2 years and wants you to identify the overall sales trend.**

    daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
    235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

    ● Explain how you would apply the Central Limit Theorem to estimate the average sales with a 95% confidence interval.

    ● Write the Python code to compute the mean sales and its confidence interval.

```python
import numpy as np

daily_sales = [
    220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
    235, 260, 245, 250, 225, 270, 265, 255, 250, 260
]

# Calculate mean and sample standard deviation
mean_sales = np.mean(daily_sales)
std_dev = np.std(daily_sales, ddof=1)

# Sample size
n = len(daily_sales)

# Standard error
standard_error = std_dev / np.sqrt(n)

# 95% confidence interval
margin_of_error = 1.96 * standard_error

lower = mean_sales - margin_of_error
upper = mean_sales + margin_of_error

print("Mean Sales:", mean_sales)
print("Standard Deviation:", std_dev)
print("95% Confidence Interval:", (lower, upper))
```

Explanation

The Central Limit Theorem (CLT) tells us that for a sufficiently large sample, the sampling distribution of the sample mean is approximately normal. We can use this to estimate the population's average sales.

For a 95% confidence interval:

    Calculate the sample mean.
    Calculate the sample standard deviation.
    Calculate the standard error:
    SE = s / √n
    Use the 95% critical value 1.96.
    Calculate:
    CI = mean ± 1.96 × SE

For the given data, the mean sales are approximately 249.75.
